## 12. Comparing Reporting Periods

The dataset contains complaint figures for multiple reporting periods. After examining the first period, I will extract the next period using the same structure so that complaint volumes and resolution patterns can eventually be compared over time.

```python
period2 = df.iloc[36:46, 0:6].copy()

period2
```

### Cleaning the Second Reporting Period

The second reporting block covers 1 April 2021 to 31 March 2022. I will apply the same cleaning process used for the first reporting period so that the two periods can be compared consistently.

```python
period2.columns = [
    "category",
    "old_pending",
    "registered",
    "total_complaints",
    "resolved",
    "under_proceedings"
]

period2 = period2.iloc[1:].copy()

period2 = period2[
    period2["category"].str.strip() != "Total"
].copy()

numeric_columns = [
    "old_pending",
    "registered",
    "total_complaints",
    "resolved",
    "under_proceedings"
]

for col in numeric_columns:
    period2[col] = pd.to_numeric(period2[col], errors="coerce")

period2
```

### Removing Empty Categories

Pending court claims and dowry claims contain zero complaints in this reporting period. These categories are excluded from the analysis because they contain no recorded cases.

```python
period2 = period2[
    period2["total_complaints"] > 0
].copy()

period2
```

## 13. Overall Complaint Outcomes

I will calculate the total number of complaints, resolved complaints and complaints under proceedings for the second reporting period.

```python
total_complaints_2 = period2["total_complaints"].sum()
total_resolved_2 = period2["resolved"].sum()
total_proceedings_2 = period2["under_proceedings"].sum()

resolution_rate_2 = (
    total_resolved_2 / total_complaints_2 * 100
)

proceeding_rate_2 = (
    total_proceedings_2 / total_complaints_2 * 100
)

print("Total complaints:", total_complaints_2)
print("Total resolved:", total_resolved_2)
print("Total under proceedings:", total_proceedings_2)
print("Resolution rate:", resolution_rate_2)
print("Proceeding rate:", proceeding_rate_2)
```

## 14. Comparing Complaint Outcomes Across Periods

Overall resolution rates changed substantially between the two reporting periods. I will now compare individual complaint categories to determine whether the change was broad-based or concentrated in particular types of complaints.

```python
comparison = period1[
    ["category", "total_complaints", "resolved", "under_proceedings"]
].merge(
    period2[
        ["category", "total_complaints", "resolved", "under_proceedings"]
    ],
    on="category",
    suffixes=("_period1", "_period2")
)

comparison
```

```python
comparison["resolution_rate_period1"] = (
    comparison["resolved_period1"] /
    comparison["total_complaints_period1"] * 100
)

comparison["resolution_rate_period2"] = (
    comparison["resolved_period2"] /
    comparison["total_complaints_period2"] * 100
)

comparison[
    [
        "category",
        "resolution_rate_period1",
        "resolution_rate_period2"
    ]
]
```

## 15. Change in Complaint Volume

Resolution rates can be affected by the length of the reporting period. I will therefore compare the number of complaints registered in each period alongside the resolution figures.

```python
print("Period 1 registered:", period1["registered"].sum())
print("Period 2 registered:", period2["registered"].sum())

change = (
    (period2["registered"].sum() - period1["registered"].sum())
    / period1["registered"].sum()
) * 100

print("Change in registered complaints:", change, "%")
```

## 16. Combining the Reporting Periods

The cleaned reporting periods are combined into a single dataset. A period column is added so that complaint categories and outcomes can be compared across time.

```python
period1["period"] = "Jan-Mar 2021"
period2["period"] = "Apr 2021-Mar 2022"

combined = pd.concat(
    [period1, period2],
    ignore_index=True
)

combined
```

```python
combined.shape
```

```python
combined[
    [
        "period",
        "category",
        "total_complaints",
        "resolved",
        "under_proceedings"
    ]
]
```

## 17. Resolution Rate Across Reporting Periods

To compare outcomes across reporting periods, I calculate the percentage of complaints in each category that were resolved.

```python
combined["resolution_rate"] = (
    combined["resolved"] / combined["total_complaints"] * 100
)

combined[
    [
        "period",
        "category",
        "total_complaints",
        "resolved",
        "resolution_rate"
    ]
]
```

### Change in Resolution Rate

I compare the resolution rate between the two reporting periods for each complaint category to identify where the largest changes occurred.

```python
rate_change = combined.pivot(
    index="category",
    columns="period",
    values="resolution_rate"
)

rate_change["change_in_percentage_points"] = (
    rate_change["Apr 2021-Mar 2022"]
    - rate_change["Jan-Mar 2021"]
)

rate_change.sort_values(
    "change_in_percentage_points",
    ascending=False
)
```

## 18. Key Finding

Resolution rates increased across all six complaint categories between the two reporting periods. However, sexual harassment at the workplace recorded the smallest improvement and remained the category with the lowest resolution rate in the second period.

```python
period2["pending_share"] = (
    period2["under_proceedings"]
    / period2["under_proceedings"].sum()
    * 100
)

period2[
    ["category", "under_proceedings", "pending_share"]
].sort_values(
    "under_proceedings",
    ascending=False
)
```

## 19. Where Are Pending Cases Concentrated?

The distribution of complaints under proceedings shows which categories contribute most to the commission's remaining caseload. This helps identify where unresolved cases are concentrated.

```python
period2[
    ["category", "under_proceedings", "pending_share"]
].sort_values(
    "pending_share",
    ascending=False
)
```

## 20. Key Findings

The analysis of the Maharashtra State Women's Commission complaint data shows three main patterns:

1. A total of 15,892 complaints were recorded in the first reporting period, of which 1,458 were resolved and 14,434 remained under proceedings.

2. In the second reporting period, 16,012 complaints were recorded, with 10,172 resolved and 5,840 under proceedings.

3. Marital problems were the largest complaint category in both periods. In the second period, they accounted for 2,352 complaints under proceedings, or about 40% of the total pending caseload.

4. Resolution rates increased across all six complaint categories between the two reporting periods. However, sexual harassment at the workplace recorded the lowest resolution rate in the second period, at 28.6%.

5. The three largest complaint categories — marital problems, social problems/rape etc., and others — accounted for about 85% of complaints in the first reporting period.

## 22. Data Limitations

The source spreadsheet contains multiple reporting periods and includes cumulative complaint outcomes such as resolved complaints and complaints under proceedings. The reporting periods are also of different lengths, so raw complaint and resolution counts should not be directly compared without considering the period covered.

The analysis therefore focuses primarily on percentages and category-level patterns. The figures in the original total rows were also checked against the individual category values, and calculations in this notebook use the category-level data.
